```python
address = '/workspaces/python-for-data-science-and-machine-learning-essential-training-part-1-3006708/data/mtcars.csv'

cars = pd.read_csv(address)

cars.columns = ['car_names','mpg','cyl','disp', 'hp', 'drat', 'wt', 'qsec', 'vs', 'am', 'gear', 'carb']

cars.head()
```

```python
groups = cars.groupby(['cyl'])
groups.sum(numeric_only=True)
```

```python
groups = cars.groupby(['car_names'])
groups.mean()
```

```python
cars.groupby("cyl").max()
```

```python
# Map AVG
cyl = cars.groupby("cyl")['hp'].mean(numeric_only=True)
cyl
```

```python
cars['hp_avg'] = cars['cyl'].map(cyl)
cars.head(1)
```

```python
cars['avg'] = cars.groupby("cyl")['hp'].transform('mean')
cars.head(1)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from pandas import DataFrame
```

```python
data = {'names':['steve', 'john', 'richard', 'sarah', 'randy', 'micheal', 'julie'],
        'age':[20, 22, 20, 21, 24, 23, 22],
        'gender':['Male', 'Male', 'Male', 'Female', 'Male', 'Male', 'Female'],
        'rank':[2, 1, 4, 5, 3, 7, 6]}

df=DataFrame(data)
df
```

```python
plt.bar(df.names,df.age)
plt.xlabel('Names')
plt.ylabel('Ages')
plt.title('Comparing Ages')
plt.show()
```

```python
plot = sns.barplot(data = df, x='names',y='age')
plot.set_title('Comparing Ages')
plot.set_xlabel('Names')
plt.show()
```

```python
plt.plot(df.names,df.age)
plt.xlabel('Names')
plt.ylabel('Ages')
plt.title('Comparing Ages')
plt.show()
```

```python
plot = sns.lineplot(data = df, x='names',y='age')
plot.set_title('Comparing Ages')
plot.set_xlabel('Names')
plt.show()
```

```python
plt.pie(df.age,labels = df.names)
plt.show()
```

```python
colors = sns.color_palette('pastel')[0:5]
plt.pie(df.age,labels = df.names, colors = colors)
plt.show()
```
